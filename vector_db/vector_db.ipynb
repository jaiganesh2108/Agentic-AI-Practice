```python
import chromadb

client = chromadb.Client()
```

```python
collection = client.create_collection("news_articles")
```

```python
collection.add(
    ids = ["id1", "id2", "id3", "id4"],
    documents = [
        "Apple is leading in the smart phone game with iphone sales up by 35%",
        "Tesla booked a major profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C and various antioxidants",
        "SpaceX got NASA contract worth 10 billion $"
    ]
)
```

```python
data = collection.get(
    includes = ["documents", "embeddings", "metadatas"]
)

for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text:", data["documents"][i])
    print("Embedding Length:", len(emb))
    print("First 10 values:", emb[:10])
```
