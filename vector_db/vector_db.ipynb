```python
import chromadb

# Create Chroma client
client = chromadb.Client()
```

```python

# Create collection
collection = client.create_collection("news_articles")
```

```python
# Add documents
collection.add(
    ids=["id1", "id2", "id3", "id4"],
    documents=[
        "Apple is leading in the smartphone game with iPhone sales up by 35%",
        "Tesla booked a major profit of 1 billion $ in Q2",
        "Apples are high in fiber, vitamin C and various antioxidants",
        "SpaceX got a NASA contract worth 10 billion $"
    ]
)
```

```python
# Retrieve data
data = collection.get(
    include=["documents", "embeddings", "metadatas"]
)

# Print results
for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i + 1}")
    print("Text:", data["documents"][i])
    print("Embedding Length:", len(emb))
    print("First 10 values:", emb[:10])
```

```python
results = collection.query(
    query_texts=["This is a query related to Elon Musk"],
    n_results = 2
)
print(results)
```
